10. **行数のカウント**

ファイルの行数をカウントせよ。確認にはwcコマンドを用いよ。

```python
with open("popular-names.txt", encoding = "utf-8") as f:
  lines = f.readlines()
print(len(lines))
```

11. **先頭からN行を出力**

ファイルの先頭N行だけを表示せよ。例えば、N=10として先頭10行を表示せよ。確認にはheadコマンドを用いよ。

```python
def n_print(file, n):
    with open(file, encoding="utf-8") as f:
        lines = f.readlines()
    for i in lines[:n]:
        print(i, end="")

n_print("popular-names.txt", 10)
```

```python
!head popular-names.txt
```

12. **末尾のN行を出力**

ファイルの末尾N行だけを表示せよ。例えば、N=10として末尾10行を表示せよ。確認にはtailコマンドを用いよ。

```python
def n_print_rev(file, n):
    with open(file, encoding="utf-8") as f:
        lines = f.readlines()
    for line in lines[-n:]:
        print(line, end="")

n_print_rev("popular-names.txt", 10)
```

```python
!tail popular-names.txt
```

13. **タブをスペースに置換**

ファイルの先頭10行に対して、タブ1文字につきスペース1文字に置換して出力せよ。確認にはsedコマンド、trコマンド、もしくはexpandコマンドなどを用いよ。

```python
with open("popular-names.txt", encoding="utf-8") as f:
    lines = f.readlines()
    for line in lines[:10]:
        line = line.replace("\t", " ")
        print(line, end="")
```

```python
!head -n 10 popular-names.txt | tr '\t' ' '
```

14. **1列目を出力**

ファイルの先頭10行に対して、各行の1列目だけを抜き出して表示せよ。確認にはcutコマンドなどを用いよ。

```python
with open("popular-names.txt", encoding="utf-8") as f:
    lines = f.readlines()
    for line in lines[:10]:
        sp = line.split("\t")
        print(sp[0])
```

```python
!head -n 10 popular-names.txt | cut -f 1
```
