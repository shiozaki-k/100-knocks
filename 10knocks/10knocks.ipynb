00. パタトクカシーー
**2つの文字列「パトカー」と「タクシー」の文字を先頭から交互に連結し、文字列「パタトクカシーー」を得よ。**

```python
a = "パトカー"
b =  "タクシー"
result = ""

for x, y in zip(a, b):
  result = result + x + y

print(result)
```

```python
result = "".join(a + b for a, b in zip("パトカー", "タクシー"))
print(result)
```

01. タクシー
**文字列「パタトクカシーー」の2, 4, 6, 8文字目を取り出し、それらを連結した文字列を得よ**

```python
a = "パタトクカシーー"
result = ""

for i in range(len(a)):
  if i % 2 == 1:
    result = result + a[i]

print(result)
```

```python
a = "パタトクカシーー"
result = ""

result = a[1::2]

print(result)
```

02. 文字列の逆順
**文字列”stressed”の文字を逆に（末尾から先頭に向かって）並べた文字列を得よ。**

```python
a = "stressed"

result = a[::-1]

print(result)
```

03. 円周率
**“Now I need a drink, alcoholic of course, after the heavy lectures involving quantum mechanics.”という文を単語に分解し、各単語の（アルファベットの）文字数を先頭から出現順に並べたリストを作成せよ**

```python
a = "Now I need a drink, alcoholic of course, after the heavy lectures involving quantum mechanics."
result = []

a = a.replace(",", "").replace(".", "")
b = a.split()
for i in range(len(b)):
  result.append(len(b[i]))

print(result)
```

04. 元素記号
**“Hi He Lied Because Boron Could Not Oxidize Fluorine. New Nations Might Also Sign Peace Security Clause. Arthur King Can.”という文を単語に分解し、1, 5, 6, 7, 8, 9, 15, 16, 19番目の単語は先頭の1文字、それ以外の単語は先頭の2文字を取り出し、取り出した文字列から単語の位置（先頭から何番目の単語か）への連想配列（辞書型もしくはマップ型）を作成せよ。**

```python
a = "Hi He Lied Because Boron Could Not Oxidize Fluorine. New Nations Might Also Sign Peace Security Clause. Arthur King Can."
x = [1, 5, 6, 7, 8, 9, 15, 16, 19]
result = {}

a = a.replace(",", "").replace(".", "")
b = a.split()

for i in range(len(b)):
  c = b[i]
  if i + 1 in x:
    result[c[0]] = i + 1
  else:
    result[c[0:2]] = i + 1

print(result)
print(result["P"])
print(result["Li"])
```

05. n-gram
**与えられたシーケンス（文字列やリストなど）からn-gramを作る関数を作成せよ。この関数を用い、”I am an NLPer”という文から文字tri-gram、単語bi-gramを得よ。**

```python
def n_gram(n, seq):
  result = []

  for i in range(len(seq) - n + 1):
      result.append(seq[i : i + n])

  return result
```

```python
a = "I am an NLPer"

char_tri_gram = n_gram(3, a)
word_bi_gram = n_gram(2, a.split())

print(char_tri_gram)
print(word_bi_gram)
```
